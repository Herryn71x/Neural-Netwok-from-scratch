```python
X = np.array([1.5, 2.0, 0.5])
W = np.array([
    [0.2, 0.8, -0.5],
    [0.5, -0.1, 0.3]
    ])
b = np.array([2.0, 0.5])
Z = np.dot(W, X) + b
def ReLU(Z):
    return np.maximum(0, Z)
A = ReLU(Z)
print(Z)
print(A)
```

```python
W2 = np.array([[0.1, -0.2],
            [-0.3, 0.4]])
b2 = np.array([0.1, 0.2])
Z2 = np.dot(W2, A) + b2

def softmax(Z):
    exp_Z = np.exp(Z)
    return exp_Z / np.sum(exp_Z)
A2 = softmax(Z2)
print(A2)
#output sum is equal to 1 so that's what softmax function do, layer 1 65% sure and layer 2 35% sure. 
```

```python
Y = np.array ([0, 1]) 
error = A2 - Y
print(Y)
print(error)
```

```python
dZ2 = A2 - Y
DW2 = np.outer(dZ2, A)
db2 = dZ2
print("Change needed for b2:", db2)
print("Change needed for W2:", DW2)
```

```python
dZ1 = np.dot(W2.T, dZ2) * (Z > 0)
dW1 = np.outer(dZ1, X)
db1 = dZ1 
print(dZ1)
print(dW1)
print(db1)
```

```python
learning_rate = 0.1
W = W - learning_rate * dW1
b = b - learning_rate * db1
print("Updated W:", W)

W2 = W2 - learning_rate * DW2
b2 = b2 - learning_rate * db2
print("Updated W2:", W2)
```

```python
epochs = 50

for i in range(epochs):
    Z = np.dot(W, X) + b
    A = np.maximum(0, Z)
    
    Z2 = np.dot(W2, A) + b2
    A2 = softmax(Z2)
    
    dZ2 = A2 - Y
    DW2 = np.outer(dZ2, A)
    db2 = dZ2
    dZ1 = np.dot(W2.T, dZ2) * (Z > 0)
    dW1 = np.outer(dZ1, X)
    db1 = dZ1
    
    W = W - learning_rate * dW1
    b = b - learning_rate * db1
    W2 = W2 - learning_rate * DW2
    b2 = b2 - learning_rate * db2
    
    if i % 5 == 0:
        print(f"Epoch {i} - Network Prediction: {A2}")
        
print(f"\nFinal Prediction after {epochs} Epochs: {A2}")
```
